STUDENT GRADE AND MARK CALCULATOR

```python
name= input("Enter your name:")
py_mark=int(input("Enter your python mark:"))
java_mark=int(input("Enter your java mark:"))
c_mark=int(input("Enter your c mark:"))
print("--------STUDENT RESULT---------")
print("Name:",name)
print("Python mark:",py_mark)
print("java mark:",java_mark)
print("c mark:",c_mark)
total= py_mark+java_mark+c_mark
avg=total/3
print("Total:",total)
print("Average:",avg)
if(avg>=35):
    print("Result: PASS")
    if(avg>=90):
        print("Grade: A")
    elif(avg>=80 and avg<90):
        print("Grade: B")
    elif(avg>=70 and avg<80):
        print("Grade: C")
    elif(avg>=55 and avg<70):
        print("Grade : D")
    elif(avg>=35 and avg<55):
        print("Grade : E")
else:
    print("Result: Fail")
    print("Grade: F")
```
