login system:
max 3 attempts
password correct print Login Successful
otherwise after 3 failures Account locked

```python
pw=1234
p=int(input("Enter your password:"))
count=1
while(count<3):
  if(pw==p):
    print("Login Successful")
    break
  else:
    p=int(input("Enter your correct password:"))
    count=count+1
if(count==3):
  print("Account Locked")
```
